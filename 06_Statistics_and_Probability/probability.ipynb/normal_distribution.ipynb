```python
import matplotlib.pyplot as plt
normal_data=np.random.normal(0,1,1000)
plt.figure(figsize=(12,6))
plt.hist(normal_data, bins=30, density=True, alpha=0.7, color='blue')
plt.title=("normal_distribution")
plt.xlabel("value")
plt.ylabel("density")
plt.show()
```
