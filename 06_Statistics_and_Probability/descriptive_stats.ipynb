```python
import pandas as pd
import numpy as np
import statistics as stats
```

```python
data=[10, 12, 14, 15, 15, 16, 18, 20, 22, 25]
```

```python
## USING NUMPY CALCULATION
import numpy as np
mean1=np.mean(data)
median1 =np.median(data)
variance=np.var(data)
sd=np.std(data)
print(mean1)
print(sd)
print(median1)
print(variance)
```

```python
## FROM STATISTICS
import statistics as stats
sm=stats.mean(data)
sm2=stats.median(data)
sm3=stats.mode(data)
var1=stats.variance(data)
sd1=stats.stdev(data)
print(sm)
print(sm2)
print(sm3)
print(var1)
print(sd1)
```

## SKEWNESS AND KURTOSIS

```python
from scipy.stats import skew,kurtosis
```

```python
normal_data=np.random.normal(0,1,1000)
left_skew=-np.random.gamma(2,1,2000)
right_skew=np.random.gamma(2,1,9990)
```

```python
print(skew(normal_data))
print(skew(left_skew))
print(skew(right_skew))
```

```python
print(kurtosis(normal_data))
print(kurtosis(left_skew))
print(kurtosis(right_skew))
```
