# 1. What is the latest price?
# 2. What was the price 5 days ago?
# 3. Compute the 5-day simple moving average.

```python
import yfinance as yf

# Download 1 month of closing prices for RELIANCE.NS
data = yf.download("RELIANCE.NS", period="1mo")["Close"]

# 1. Latest price
latest_price = data.iloc[-1]
print("Latest Price:", latest_price)

# 2. Price 5 trading days ago
price_5_days_ago = data.iloc[-6]
print("Price 5 Days Ago:", price_5_days_ago)

# 3. 5-day Simple Moving Average
sma_5 = data.rolling(window=5).mean()

print("Latest 5-Day SMA:", sma_5.iloc[-1])
```
